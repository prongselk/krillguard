```python
start = Point(-54.241944, -36.400833)  #merton rock s georgia


bearing = 44  #North - East
dist = 5.3 * 1.852


destination = distance(kilometers=dist).destination(point=start, bearing=bearing)


print(f"Destination: {destination.latitude}, {destination.longitude}")
```
